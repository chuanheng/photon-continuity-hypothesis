```python
import json
import sys
import unittest
from io import StringIO
from pathlib import Path

here = Path.cwd().resolve()
root = next(path for path in (here, *here.parents)
            if (path / "research_state.json").is_file())
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
```

```python
from v422_v431.model import run

result = json.loads(json.dumps(run(), allow_nan=False))
reference = json.loads((root / "v422_v431" / "results.json").read_text())
assert result == reference, "运行输出和归档输出不同"
for version in ["v4.22", "v4.27", "v4.30"]:
    print(version, json.dumps(result[version], ensure_ascii=False, indent=2))
```

```python
suite = unittest.defaultTestLoader.loadTestsFromName("v422_v431.test_model")
log = StringIO()
tests = unittest.TextTestRunner(stream=log).run(suite)
if not tests.wasSuccessful():
    print(log.getvalue())
assert tests.wasSuccessful() and tests.testsRun >= 10
print(f"{tests.testsRun} 项测试通过")
```
