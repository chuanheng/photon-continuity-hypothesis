# v4.32–v4.41 量子因果切断与校准

从仓库运行。需要 Python、NumPy 和 SciPy。全部结果来自理论模型，无光学实测。

```python
import json
import sys
import unittest
from io import StringIO
from pathlib import Path

here = Path.cwd().resolve()
root = next(path for path in (here, *here.parents)
            if (path / "research_state.json").is_file())
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
```

```python
from v432_v441.model import run

result = json.loads(json.dumps(run(), allow_nan=False))
reference = json.loads((root / "v432_v441" / "results.json").read_text())
assert result == reference, "运行输出和归档输出不同"
for version in ["v4.34", "v4.37", "v4.38", "v4.40", "v4.41"]:
    print(version, json.dumps(result[version], ensure_ascii=False, indent=2))
```

```python
suite = unittest.defaultTestLoader.loadTestsFromName("v432_v441.test_model")
log = StringIO()
tests = unittest.TextTestRunner(stream=log).run(suite)
if not tests.wasSuccessful():
    print(log.getvalue())
assert tests.wasSuccessful() and tests.testsRun == 10
print(f"{tests.testsRun} 项测试通过")
```
